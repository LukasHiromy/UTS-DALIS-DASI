## Perform and interpret the additional test

### Subtask:
Perform a Kruskal-Wallis test to compare the distribution of 'percent' across different 'race_eth_name' groups.


**Reasoning**:
Perform the Kruskal-Wallis test to compare the distribution of 'percent' across different 'race_eth_name' groups and print the results.



```python
from scipy.stats import kruskal

# Create a list of data arrays for each race_eth_name group
race_eth_groups = [df[df['race_eth_name'] == group]['percent'].dropna() for group in df['race_eth_name'].unique()]

# Remove empty arrays from the list
race_eth_groups = [group for group in race_eth_groups if not group.empty]

# Perform the Kruskal-Wallis test
kruskal_statistic, p_value = kruskal(*race_eth_groups)

# Print the results
print(f"Kruskal-Wallis Test Statistic: {kruskal_statistic}")
print(f"P-value: {p_value}")

# Interpret the results
alpha = 0.05
print("\nInterpretation:")
if p_value < alpha:
    print(f"Since the p-value ({p_value:.4f}) is less than the significance level ({alpha}), we reject the null hypothesis.")
    print("There is a statistically significant difference in the distribution of 'percent' among the different 'race_eth_name' groups.")
else:
    print(f"Since the p-value ({p_value:.4f}) is greater than the significance level ({alpha}), we fail to reject the null hypothesis.")
    print("There is no statistically significant difference in the distribution of 'percent' among the different 'race_eth_name' groups.")
```

**Reasoning**:
Interpret the results of the Kruskal-Wallis test, including the p-value and test statistic, and discuss its relevance to the dataset.



```python
print("\nRelevance to the dataset:")
print("The Kruskal-Wallis test revealed a statistically significant difference (p = {:.6f}) in the distribution of the percentage of residents using different modes of transportation across different race/ethnicity groups.".format(p_value))
print("The test statistic (H = {:.2f}) quantifies the degree of difference in the ranks of 'percent' values across the groups. A larger value indicates greater differences between the groups.")
print("This finding is relevant as it suggests that race and ethnicity are associated with variations in transportation mode usage. Further analysis, such as post-hoc tests, could be conducted to identify which specific race/ethnicity groups differ significantly from each other.")
print("This information can be valuable for targeted transportation planning and policy development to address potential disparities in access to and usage of different transportation modes among different racial and ethnic communities.")
```
